DATA CLEARNING START FROM HERE.

```python
# Never modify the raw dataset directly
df_clean = df.copy()

print("Raw dataset shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
```

```python
# Missing children values are treated as zero
# because children represents a count of children associated with the booking.

df_clean["children"] = df_clean["children"].fillna(0)

print("Missing children values after cleaning:",
      df_clean["children"].isnull().sum())
```

```python
# Missing country values are labelled as Unknown.
# We do not guess a country for these bookings.

df_clean["country"] = df_clean["country"].fillna("Unknown")

print("Missing country values after cleaning:",
      df_clean["country"].isnull().sum())
```

```python
# Agent and company are identifier fields.
# Missing values are kept as missing because assigning an artificial
# ID such as 0 could incorrectly imply that a real agent/company exists.

df_clean["agent"] = df_clean["agent"].astype("Int64")
df_clean["company"] = df_clean["company"].astype("Int64")

print("Missing agent values:", df_clean["agent"].isnull().sum())
print("Missing company values:", df_clean["company"].isnull().sum())
```

```python
df_clean["reservation_status_date"] = pd.to_datetime(
    df_clean["reservation_status_date"],
    errors="coerce"
)

print(df_clean["reservation_status_date"].dtype)
```

```python
print("Missing/Invalid dates:", df_clean["reservation_status_date"].isna().sum())
print("Total rows:", len(df_clean))
```

```python
integer_columns = [
    "is_canceled",
    "lead_time",
    "arrival_date_year",
    "arrival_date_week_number",
    "arrival_date_day_of_month",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "children",
    "babies",
    "is_repeated_guest",
    "previous_cancellations",
    "previous_bookings_not_canceled",
    "booking_changes",
    "days_in_waiting_list",
    "required_car_parking_spaces",
    "total_of_special_requests"
]

for column in integer_columns:
    df_clean[column] = df_clean[column].astype("Int64")
```

```python
df_clean["adr"] = pd.to_numeric(
    df_clean["adr"],
    errors="coerce"
)
```

```python
text_columns = [
    "hotel",
    "meal",
    "country",
    "market_segment",
    "distribution_channel",
    "reserved_room_type",
    "assigned_room_type",
    "deposit_type",
    "customer_type",
    "reservation_status"
]

for column in text_columns:
    df_clean[column] = df_clean[column].astype("string").str.strip()
```

# Assumptions and Data Preparation Decisions

## 1. Raw data preservation

The original dataset is treated as raw source data and is not modified directly. All preparation steps are performed on a copy of the dataset.

## 2. Children

The `children` column represents the number of children associated with a booking.

Only four records contain missing values. These missing values are treated as zero because the absence of a recorded child count is interpreted as no children being associated with that booking.

## 3. Country

Missing values in `country` are labelled as `Unknown`.

A country is not inferred or replaced with the most common country because doing so could introduce incorrect information into the dataset.

## 4. Agent

The `agent` column represents an identifier associated with the booking agent.

Missing values are retained as missing rather than replacing them with an artificial identifier such as 0.

## 5. Company

The `company` column represents an identifier associated with the company making the booking.

Missing values are retained as missing rather than replacing them with an artificial identifier such as 0.

## 6. Date type

`reservation_status_date` is treated as a date field and converted from text to Pandas datetime.

The date values are interpreted using the day-month-year format represented in the source dataset.

## 7. Numeric fields

Fields representing counts, flags, durations, and rates are maintained as numeric data types so that they can be used correctly in calculations and future analysis.

## 8. Categorical fields

Fields such as `hotel`, `meal`, `market_segment`, `distribution_channel`, `deposit_type`, and `customer_type` are treated as categorical/text fields.

## 9. Duplicate records

Exact duplicate rows are considered duplicate records and are removed.

Rows are not considered duplicates merely because they share the same hotel, customer characteristics, or other individual field values.

## 10. Negative values

Negative values in fields where negative values are logically impossible, such as lead time or average daily rate, are treated as data-quality issues and are checked separately.

## 11. Zero-guest bookings

Bookings with zero adults, children, and babies are flagged for investigation rather than automatically removed during the initial preparation stage.

## 12. Scope of Task 1

This task focuses on data preparation and data quality.

No predictive model, statistical conclusion, or business performance claim is made at this stage.

The cleaned dataset is prepared for subsequent exploratory analysis and modelling tasks.
